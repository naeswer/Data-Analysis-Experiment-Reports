```python
import chardet
import pandas as pd

# 读取文件二进制内容，检测编码
file_path = r"C:\Users\win11\Downloads\Pokemon.csv"
with open(file_path, "rb") as f:
    raw_data = f.read()
    result = chardet.detect(raw_data)  # 检测编码
    encoding = result["encoding"]  # 获取检测到的编码
    confidence = result["confidence"]  # 检测置信度（越接近1越准确）

print(f"检测到文件编码：{encoding}，置信度：{confidence}")

# 用检测到的编码读取文件
data = pd.read_csv(file_path, encoding=encoding)
data
```
